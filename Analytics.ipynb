```python
# Plot age distribution
sns.set_theme(style='white')
sns.distplot(df.age, axlabel="Age", color="dimgrey")
plt.axvline(x=df.age.mean(), color='red', linestyle='--', label=f'Mean: {df.age.mean():.2f}', linewidth=2)
plt.text(df.age.mean() + 0.5, 0.01, "Mean", rotation=90)
plt.axvline(x=df.age.median(), color='blue', linestyle='--', label=f'Median: {df.age.median():.2f}', linewidth=2)
plt.text(df.age.median() - 3, 0.01, "Median", rotation=90)
plt.title("Age Distribution")

plt.legend(facecolor='white', framealpha=1)
```

## Wilcoxon Rank Sum Test

```python
from scipy.stats import ranksums

# Wilcoxon Rank Sum Test of each treatment
for treatment in range(1, 7):
    print("treatment ", treatment)
    print(
        ranksums(
            df.buttonpresses[df.treat == treatment],
            df.buttonpresses[df.treat == 7],
            alternative="greater",
        )
    )
```

```python
# Create a table (DataFrame) of rank sum tests
index_list = ["T" + str(i) for i in range(1, 8)]
wilcoxon_df = pd.DataFrame(index=index_list, columns=index_list)

for treatment in range(1, 8):
    for treatment_comparison in range(1, 8):
        pval = ranksums(
            df.buttonpresses[df.treat == treatment],
            df.buttonpresses[df.treat == treatment_comparison],
            alternative="greater",
        )[1]
        wilcoxon_df.iloc[treatment - 1, treatment_comparison - 1] = pval
wilcoxon_df
```

The Hypothesis being test is "Row <= Column"

```python
# Convert to Latex
print(wilcoxon_df.to_latex())
```

```python
for treatment1 in range(1, 7):
    print("> Treatment ", treatment1)
    for treatment in range(1, 7):
        print("treatment ", treatment)
        print(
            ranksums(
                df.buttonpresses[df.treat == treatment],
                df.buttonpresses[df.treat == treatment1],
                alternative="greater",
            )
        )
    print("\n")
```
